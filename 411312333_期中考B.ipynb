````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
DB_PATH = "cafe.db"                     # SQLite 資料庫檔名
MENU_INIT = [("美式咖啡", 50, 5), ("拿鐵", 65, 4), ("抹茶歐蕾", 70, 3)]  # (品項, 單價, 庫存)
ITEMS = [m[0] for m in MENU_INIT]       # 允許的品項清單
MAX_QTY = 3                             # 單次最多杯數（1～3）
DAILY_LIMIT = 3                         # 每人每日累計上限


def init_db():
    """刪除舊資料表 → 建立新資料表 → 放入初始資料"""
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        cur.execute("DROP TABLE IF EXISTS orders")
        cur.execute("DROP TABLE IF EXISTS menu")
        # 菜單表：stock 是「目前剩餘」，initial_stock 是「今日起始庫存」
        cur.execute("""CREATE TABLE menu (
            item TEXT PRIMARY KEY,
            price INTEGER NOT NULL,
            initial_stock INTEGER NOT NULL,
            stock INTEGER NOT NULL)""")
        # 訂單表：status 只有「已訂購」與「已取消」
        cur.execute("""CREATE TABLE orders (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student TEXT NOT NULL,
            item TEXT NOT NULL REFERENCES menu(item),
            qty INTEGER NOT NULL,
            total INTEGER NOT NULL,
            status TEXT NOT NULL DEFAULT '已訂購')""")
        cur.executemany(
            "INSERT INTO menu (item, price, initial_stock, stock) VALUES (?, ?, ?, ?)",
            [(i, p, s, s) for i, p, s in MENU_INIT])
        conn.commit()
    finally:
        conn.close()


init_db()
print("✅ 資料庫已建立並放入初始資料")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
def show_tables():
    """把菜單與訂單兩張表印出來"""
    conn = sqlite3.connect(DB_PATH)
    try:
        print("📋 菜單 menu")
        display(pd.read_sql_query("SELECT * FROM menu", conn))
        print("🧾 訂單 orders")
        display(pd.read_sql_query("SELECT * FROM orders", conn))
    finally:
        conn.close()


show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
ACTIONS = ["訂購", "取消", "查詢", "統計"]   # 允許的動作（其餘一律視為「未知」）
_ai_cache = {}                                # 同一句話只呼叫一次 AI

SYSTEM_PROMPT = """你是咖啡廳訂餐系統的「句子解讀器」。只輸出一行，格式固定為：動作,品項,數量
規則：
1. 動作只能是：訂購、取消、查詢、統計、未知。
   - 想買、訂、來一杯 → 訂購；不要了、取消 → 取消；還有嗎、剩幾杯 → 查詢；
   - 問今天賣得怎樣、銷售、營收、統計 → 統計；看不懂或與咖啡廳無關 → 未知。
2. 品項只能是：美式咖啡、拿鐵、抹茶歐蕾、無。菜單以外的品項、或沒提到品項時輸出「無」。
3. 數量用阿拉伯數字（兩杯→2、三杯→3、五杯→5）；沒說數量時輸出 1。不要自己限制範圍。
4. 無法判斷時輸出：未知,無,1
5. 你不可以回答問題、計算金額、查詢庫存、修改資料，也不可以輸出格式以外的任何文字。"""


def call_groq(text):
    """呼叫 Groq；遇到 429 等 30 秒後重試，最多重試 3 次"""
    for attempt in range(4):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                max_completion_tokens=500,
                messages=[{"role": "system", "content": SYSTEM_PROMPT},
                          {"role": "user", "content": text}],
            )
            time.sleep(2)                     # 每次呼叫後休息 2 秒，避免太快
            return (resp.choices[0].message.content or "").strip()
        except Exception as e:
            msg = str(e)
            if ("429" in msg or "rate" in msg.lower()) and attempt < 3:
                print("⏳ 超過使用量，等待 30 秒後重試…")
                time.sleep(30)
                continue
            print("❌ AI 呼叫失敗：", type(e).__name__)   # 只印錯誤類型，避免洩漏資訊
            return ""
    return ""


def parse_intent(text):
    """把一句話交給 AI，再用程式檢查，回傳 (動作, 品項, 數量, 原始回覆)"""
    text = (text or "").strip()
    if not text:
        return ("未知", "無", 1, "")
    if text in _ai_cache:
        return _ai_cache[text]

    raw = call_groq(text)
    line = raw.splitlines()[0] if raw else ""          # 只看第一行
    parts = [p.strip() for p in re.split(r"[,，]", line)]

    # 動作：只在第 1 段找允許清單的值，找不到 → 未知
    action = "未知"
    for a in ACTIONS:
        if parts and a in parts[0]:
            action = a
            break

    # 品項：在整行找允許清單的值，找不到 → 無
    item = "無"
    for it in ITEMS:
        if it in line:
            item = it
            break

    # 數量：從第 3 段用正規表示式找整數，找不到 → 1（範圍由功能函式檢查）
    qty = 1
    if len(parts) >= 3:
        m = re.search(r"-?\d+", parts[2])
        if m:
            qty = int(m.group())

    result = (action, item, qty, raw)
    _ai_cache[text] = result
    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
ai_tests = [
    ("我要一杯拿鐵",         "訂購", "拿鐵",     1),     # 正常
    ("幫我訂兩杯美式咖啡",   "訂購", "美式咖啡", 2),     # 正常
    ("取消我的抹茶歐蕾",     "取消", "抹茶歐蕾", "*"),   # 正常
    ("拿鐵還有剩嗎",         "查詢", "拿鐵",     "*"),   # 正常
    ("今天各品項賣得怎樣",   "統計", "*",        "*"),   # 正常
    ("我要訂3杯抹茶歐蕾",    "訂購", "抹茶歐蕾", 3),     # 邊界：上限
    ("我要訂五杯拿鐵",       "訂購", "拿鐵",     5),     # 邊界：超過上限（由程式拒絕）
    ("幫我訂拿鐵",           "訂購", "拿鐵",     1),     # 邊界：沒說數量視為 1
    ("我要一杯珍珠奶茶",     "*",    "無",       "*"),   # 錯誤：菜單沒有
    ("今天天氣好好",         "未知", "無",       "*"),   # 錯誤：無關
]


def run_ai_tests():
    passed = 0
    for i, (text, ea, ei, eq) in enumerate(ai_tests, 1):
        action, item, qty, raw = parse_intent(text)
        ok = (ea in ("*", action)) and (ei in ("*", item)) and (eq in ("*", qty))
        passed += ok
        print(f"{'✅' if ok else '❌'} #{i} 「{text}」")
        print(f"     AI 原始回覆：{raw!r}")
        print(f"     系統解讀：動作={action}、品項={item}、數量={qty}（預期：{ea}、{ei}、{eq}）")
    print(f"\n📊 AI 解讀測試通過 {passed} / {len(ai_tests)} 句")


run_ai_tests()
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
def get_menu_text():
    """回傳目前菜單文字（用於無法判斷時）"""
    conn = sqlite3.connect(DB_PATH)
    try:
        rows = conn.execute("SELECT item, price, stock FROM menu ORDER BY rowid").fetchall()
    finally:
        conn.close()
    return "、".join(f"{i}（{p} 元，剩 {s} 杯）" for i, p, s in rows)


def do_order(name, item, qty):
    """訂購：依序檢查 姓名 → 品項 → 數量 → 個人上限 → 庫存"""
    if not name:
        return "❗ 訂購需要姓名，請先輸入姓名。"
    if item not in ITEMS:
        return "❗ 看不出您要訂哪個品項。目前品項：" + get_menu_text()
    if not (1 <= qty <= MAX_QTY):
        return f"❗ 每次只能訂 1～{MAX_QTY} 杯，您要的是 {qty} 杯，無法訂購。"

    conn = sqlite3.connect(DB_PATH)
    try:
        row = conn.execute("SELECT price, stock FROM menu WHERE item = ?", (item,)).fetchone()
        if row is None:
            return "❗ 查無此品項。"
        price, stock = row

        # A2：該學生今天已訂的杯數（已取消不計）
        row = conn.execute(
            "SELECT COALESCE(SUM(qty), 0) FROM orders WHERE student = ? AND status = '已訂購'",
            (name,)).fetchone()
        already = row[0] if row is not None else 0
        if already + qty > DAILY_LIMIT:
            left = max(DAILY_LIMIT - already, 0)
            return f"❗ 每人今天最多訂 {DAILY_LIMIT} 杯，{name} 已訂 {already} 杯，最多還能訂 {left} 杯。"

        if stock < qty:
            return f"❗ {item} 庫存不足，目前只剩 {stock} 杯。"

        total = price * qty
        conn.execute("UPDATE menu SET stock = stock - ? WHERE item = ?", (qty, item))
        conn.execute("INSERT INTO orders (student, item, qty, total, status) VALUES (?, ?, ?, ?, '已訂購')",
                     (name, item, qty, total))
        conn.commit()
        return f"✅ {name} 訂購成功：{item} × {qty} 杯，共 {total} 元。"
    finally:
        conn.close()


def do_cancel(name, item):
    """取消：把該學生該品項「最新一筆」仍有效的訂單改為已取消，並釋出庫存"""
    if not name:
        return "❗ 取消需要姓名，請先輸入姓名。"
    if item not in ITEMS:
        return "❗ 看不出您要取消哪個品項。目前品項：" + get_menu_text()

    conn = sqlite3.connect(DB_PATH)
    try:
        row = conn.execute(
            "SELECT id, qty FROM orders WHERE student = ? AND item = ? AND status = '已訂購' "
            "ORDER BY id DESC LIMIT 1", (name, item)).fetchone()
        if row is None:
            return f"❗ 找不到 {name} 訂購「{item}」的訂單，無法取消。"
        order_id, qty = row
        conn.execute("UPDATE orders SET status = '已取消' WHERE id = ?", (order_id,))
        conn.execute("UPDATE menu SET stock = stock + ? WHERE item = ?", (qty, item))
        conn.commit()
        return f"✅ 已取消 {name} 的 {item} × {qty} 杯，庫存已釋出。"
    finally:
        conn.close()


def do_query(item):
    """查詢剩餘杯數（不需姓名）"""
    if item not in ITEMS:
        return "❗ 看不出您要查哪個品項。目前品項：" + get_menu_text()
    conn = sqlite3.connect(DB_PATH)
    try:
        row = conn.execute("SELECT stock FROM menu WHERE item = ?", (item,)).fetchone()
        if row is None:
            return "❗ 查無此品項。"
        return f"🔎 {item} 目前剩餘 {row[0]} 杯。"
    finally:
        conn.close()


def do_stats():
    """A3 統計：menu LEFT JOIN orders + GROUP BY，沒人訂的品項也會列出"""
    conn = sqlite3.connect(DB_PATH)
    try:
        df = pd.read_sql_query(
            """SELECT m.item AS 品項,
                      m.price AS 單價,
                      m.initial_stock AS 庫存,
                      COALESCE(SUM(o.qty), 0) AS 已售杯數,
                      m.initial_stock - COALESCE(SUM(o.qty), 0) AS 剩餘杯數,
                      COALESCE(SUM(o.total), 0) AS 營收
               FROM menu AS m
               LEFT JOIN orders AS o ON o.item = m.item AND o.status = '已訂購'
               GROUP BY m.item, m.price, m.initial_stock
               ORDER BY m.rowid""", conn)
    finally:
        conn.close()
    return "📈 今日銷售統計（營收合計 " + str(int(df["營收"].sum())) + " 元）\n" + df.to_string(index=False)
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
def handle_request(text, name):
    """主流程：AI 解讀 → 依動作呼叫對應功能 → 回覆文字"""
    text = (text or "").strip()
    name = (name or "").strip()
    if not text:
        return "請輸入您的需求。"
    action, item, qty, raw = parse_intent(text)
    if action == "訂購":
        return do_order(name, item, qty)
    if action == "取消":
        return do_cancel(name, item)
    if action == "查詢":
        return do_query(item)
    if action == "統計":
        return do_stats()
    return "🤔 抱歉，我無法判斷您的需求。目前品項：" + get_menu_text()


# 流程測試：(姓名, 使用者說的話, 回覆中應包含的文字)
flow_tests = [
    ("小明", "我要一杯拿鐵",         "65 元"),        # 1 訂購成功
    ("小明", "再來兩杯拿鐵",         "130 元"),       # 2 累計 3 杯
    ("小明", "我還要一杯美式咖啡",   "最多訂 3 杯"),  # 3 超過每人上限
    ("小華", "我要訂三杯抹茶歐蕾",   "210 元"),       # 4 抹茶售完
    ("小美", "我要一杯抹茶歐蕾",     "庫存不足"),     # 5 庫存不足
    ("小明", "取消我的拿鐵",         "已取消"),       # 6 取消最新一筆（2 杯）
    ("小華", "取消我的美式咖啡",     "找不到"),       # 7 查無訂單
    ("",     "我要一杯拿鐵",         "需要姓名"),     # 8 沒填姓名
    ("小美", "拿鐵還剩幾杯",         "剩餘 3 杯"),    # 9 查詢
    ("店長", "今天各品項賣得怎樣",   "營收"),         # 10 統計
]

init_db()    # 流程測試前先重建資料，確保從初始狀態開始
passed = 0
for i, (name, text, expect) in enumerate(flow_tests, 1):
    reply = handle_request(text, name)
    ok = expect in reply
    passed += ok
    print(f"{'✅' if ok else '❌'} 步驟 {i}｜姓名：{name or '（空白）'}｜「{text}」\n{reply}\n")
print(f"📊 流程測試通過 {passed} / {len(flow_tests)} 步\n")
show_tables()
````
